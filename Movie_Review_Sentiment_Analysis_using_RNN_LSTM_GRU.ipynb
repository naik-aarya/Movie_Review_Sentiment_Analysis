# Data Cleaning

```python
import nltk
nltk.download('all')
```

```python
#  Basic Cleaning , word tokenize , remove stop words
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
```

```python
data
```

```python
import re
def clean_text(text):
    """
    NLP preprocessing pipeline:
    1. Lowercase
    2. Remove HTML tags
    3. Remove URLs
    4. Remove Emojis
    5. Remove Punctuation
    6. Remove Numbers
    7. Remove Extra Whitespaces
    8. Word Tokenization
    9. Stopword Removal
    """

    # Lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove Emojis (non-ASCII characters)
    text = re.sub(r'[^\x00-\x7F]+', '', text)

    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenization
    tokens = word_tokenize(text)

    # Stopword removal
    tokens = [word for word in tokens if word not in stop_words]

    # Join tokens back into a sentence
    text = " ".join(tokens)

    return text
```

```python
data['cleaned_review']= data['review'].apply(clean_text)
```

```python
data
```

# Tokenizer

Used to convert text to number

```python
from tensorflow.keras.preprocessing.text import Tokenizer
# keeps only most frequently occured 5000 words
tokenizer = Tokenizer(num_words=5000)
# Build the vocabulary
tokenizer.fit_on_texts(data['cleaned_review'])
# convert text to sequence of integers
x = tokenizer.texts_to_sequences(data['cleaned_review'])
y = data['Sentiment']
```

```python
tokenizer.word_index # list of vocabulary
```

```python
x[0] # first sequence
```

# Split data for training and testing

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

# **Padding**

RNNs require all input sequences to have the same length.

padding='post': Adds padding values (usually 0) at the end of sequences that are shorter than the desired length.

truncating='post': Removes extra elements from the end of sequences that are longer than the desired length.

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
x_train = pad_sequences(x_train,maxlen=500,padding='post',truncating='post')
x_test = pad_sequences(x_test,maxlen=500,padding='post',truncating='post')
```

```python
x_train.shape
```

# Building RNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN , Dense , Embedding
model = Sequential()
model.add(Embedding(input_dim=5000 , output_dim=128 , input_length=500))
model.add(SimpleRNN(128))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
rnn = model.fit(x_train,y_train,epochs=10,batch_size=64,validation_split=0.2,verbose=1)
```

```python
loss, accuracy = model.evaluate(x_test, y_test)

print("Loss:", loss)
print("Accuracy:", accuracy)
```

# LSTM

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM , Dense , Embedding
model = Sequential()
model.add(Embedding(input_dim=5000 , output_dim=128 , input_length=500))
model.add(LSTM(128))
model.add(Dense(1,activation='sigmoid'))
```

# GRU

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU , Dense , Embedding
model = Sequential()
model.add(Embedding(input_dim=5000 , output_dim=128 , input_length=500))
model.add(GRU(128))
model.add(Dense(1,activation='sigmoid'))
```
